# Helmet Detection Dataset Analysis

This notebook is used to inspect and validate the helmet detection dataset before model training.

```python
import os
import yaml

with open("../data/data.yaml", "r") as file:
    data = yaml.safe_load(file)

data
```

```python
#check the classes
print("Classes:", data["names"])
```

```python
#count the images
train_images = os.listdir("../data/train/images")
valid_images = os.listdir("../data/valid/images")
test_images = os.listdir("../data/test/images")

print("Train images:", len(train_images))
print("Validation images:", len(valid_images))
print("Test images:", len(test_images))
```

```python
#count the label files
train_labels = os.listdir("../data/train/labels")
valid_labels = os.listdir("../data/valid/labels")
test_labels = os.listdir("../data/test/labels")

print("Train labels:", len(train_labels))
print("Validation labels:", len(valid_labels))
print("Test labels:", len(test_labels))
```

```python
#look at one label file
file_name = train_labels[0]
with open("../data/train/labels/" + file_name, "r") as file:
    print(file.read())
```

```python
#count Helmet vs No-Helmet to check whther the datset is balanced or not
helmet_count = 0
no_helmet_count = 0

for file_name in train_labels:
    with open("../data/train/labels/" + file_name, "r") as file:
        for line in file:
            if line.strip():
                class_id = int(line.split()[0])

                if class_id == 0:
                    helmet_count += 1

                elif class_id == 1:
                    no_helmet_count += 1

print("Helmet:", helmet_count)
print("No-Helmet:", no_helmet_count)
```

```python
helmet_count = 0
no_helmet_count = 0

for file_name in valid_labels:
    with open("../data/valid/labels/" + file_name, "r") as file:
        for line in file:
            if line.strip():
                class_id = int(line.split()[0])

                if class_id == 0:
                    helmet_count += 1
                elif class_id == 1:
                    no_helmet_count += 1

print("Helmet:", helmet_count)
print("No-Helmet:", no_helmet_count)
```

```python
helmet_count = 0
no_helmet_count = 0

for file_name in test_labels:
    with open("../data/test/labels/" + file_name, "r") as file:
        for line in file:
            if line.strip():
                class_id = int(line.split()[0])

                if class_id == 0:
                    helmet_count += 1
                elif class_id == 1:
                    no_helmet_count += 1

print("Helmet:", helmet_count)
print("No-Helmet:", no_helmet_count)
```

```python
#make sure  images can display correctly
import cv2
import matplotlib.pyplot as plt
image_name = train_images[0]

image_path = "../data/train/images/" + image_name

image = cv2.imread(image_path)
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(image)
plt.axis("off")
plt.show()
```
