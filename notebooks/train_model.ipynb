# Helmet Detection Model Training

This notebook is used to train a YOLO model on the helmet detection dataset.

The dataset contains two classes:

- Helmet
- No-Helmet

A pretrained YOLO model is used as the starting point and then trained on the custom dataset.

```python
from ultralytics import YOLO
model = YOLO("yolo26n.pt")
```

```python
# quick training test to make sure the dataset, GPU, and training setup are working correctly.
model.train(
    data="../data/data.yaml",
    epochs=3,
    imgsz=640,
    batch=2,
    workers=2,
    device=0
)
```

```python
# Load a fresh pretrained YOLO model for the longer training run

model = YOLO("yolo26n.pt")
# Train the model for 20 epochs using the same settings as the baseline

model.train(
    data="../data/data.yaml",
    epochs=100,
    imgsz=640,
    batch=2,
    workers=2,
    device=0,
    name="helmet_100_epochs"
)
```

## Fine-Tuning the Best Model

During the 100-epoch training, the model reached its best validation performance around epoch 89.

After that point, the results did not improve much and stayed at a similar level until epoch 100.

This suggests that simply training for more epochs may not give a large improvement.

So instead of running more epochs normally the best saved model will now be fine-tuned for a few more epochs using a smaller learning rate.

```python
# Load the best model from the previous training run

from ultralytics import YOLO

model = YOLO("runs/detect/helmet_100_epochs/weights/best.pt")
```

```python
# Fine-tune the best model using a smaller learning rate

model.train(
    data="../data/data.yaml",
    epochs=30,
    imgsz=640,
    batch=2,
    workers=2,
    device=0,
    optimizer="AdamW",
    lr0=0.0005,
    name="helmet_finetuned_smallLR"
)
```

## Fine-Tuning Result

The best model from the 100-epoch training was fine-tuned for another 30 epochs using a smaller learning rate.

The fine-tuned model did not improve the validation performance. Its mAP@50 and mAP@50-95 were slightly lower than the original best model.

Therefore, the original 100-epoch `best.pt` model is kept as the current best model.

Since many helmet objects in the dataset are small, the next experiment will check whether using a larger image size can improve detection.

## Testing a Larger Image Size

Many helmets in the dataset appear as small objects in the images.

The current best model was trained using an image size of 640. Before doing another training run, the same model will be evaluated using a larger image size of 768 to check whether small-object detection improves.

```python
# Load the best model from the 100-epoch training 

from ultralytics import YOLO

model = YOLO("runs/detect/helmet_100_epochs/weights/best.pt")
```

```python
# Evaluate the same model using a larger image size
results_768 = model.val(
    data="../data/data.yaml",
    imgsz=768,
    batch=1,
    device=0
)
```

## Final Model Selection

The model was already trained for 100 epochs and the validation results had mostly stopped improving near the end of training.

A smaller learning rate was also tested for another 30 epochs, but it did not improve the model. A larger image size of 768 was also evaluated, but the improvement in mAP@50 was very small and the inference became slower.

Because of this, no additional training epochs will be used.

The original 100-epoch `best.pt` model will be kept as the final model for the next stage of the project.
