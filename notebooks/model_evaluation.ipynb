# Final Model Evaluation


The original 100-epoch `best.pt` model was selected as the final model because additional fine-tuning and a larger image size did not provide a meaningful improvement.

This notebook evaluates the final model on the test dataset

```python
# Import YOLO
from ultralytics import YOLO
```

```python
# Load the best model from the 100-epoch training run

model = YOLO("runs/detect/helmet_100_epochs/weights/best.pt")
```

## Test Set Evaluation

The final model is now evaluated on the test dataset.


```python
# Evaluate the final model on the test images

test_results = model.val(
    data="../data/data.yaml",
    split="test",
    imgsz=640,
    batch=2,
    device=0,
    name="final_test"
)
```

## Final Test Results

The final model was evaluated on the test dataset, which was not used during model tuning.

The final test results were:

- Precision: 0.814
- Recall: 0.710
- mAP@50: 0.802
- mAP@50-95: 0.458

The test results were slightly lower than the validation results, which is expected because the test set contains unseen data.

The model performed better on the Helmet class than the No-Helmet class, but both classes achieved reasonable detection performance.

These results will be kept as the final evaluation of the detection model.
