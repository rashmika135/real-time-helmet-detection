# Image Prediction

The final trained model is now tested on a images.

```python

from ultralytics import YOLO
```

```python
# Load the final model selected from the training experiments
model = YOLO("runs/detect/helmet_100_epochs/weights/best.pt")
```

```python
# Run prediction on one image first

results = model.predict(
    source="../data/test/images/51560a4aeab8eaa116000005_jpeg_jpg.rf.7106690cd308c01bacebcaf9b094375c.jpg",
    imgsz=640,
    conf=0.25,
    save=True)
```

```python
results = model.predict(
    source="../data/test/images/veh_nga3muitau554_ID415_xe_may_model-_color-_86C36387_bien_unknown_15-03-2024_16-35-49_640_jpg.rf.8f2d5b6e7b6f3d6133970692d4c55998.jpg",
    imgsz=640,
    conf=0.25,
    save=True)
```

```python
results = model.predict(
    source="../data/test/images/22301_jpg.rf.13e79a67f072648bc5c569c9d1c57204.jpg",
    imgsz=640,
    conf=0.25,
    save=True)
```

```python
# Show the predicted image inside the notebook

import matplotlib.pyplot as plt

predicted_image = results[0].plot()

plt.figure(figsize=(12, 8))
plt.imshow(predicted_image[:, :, ::-1])
plt.show()
```
