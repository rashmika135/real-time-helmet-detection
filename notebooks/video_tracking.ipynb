# Helmet Tracking

The trained helmet detection model is now combined with object tracking.

The tracker gives each detected Helmet or No-Helmet object an ID and tries to keep the same ID while the object moves through the video.


```python
from ultralytics import YOLO
```

```python
# Load the best model from the 100-epoch training run

model = YOLO("runs/detect/helmet_100_epochs/weights/best.pt")
```

```python
# Set the path of the video to track
video_path = "../data/video/testing_full_video.mp4"
```

## Run Object Tracking

ByteTrack is used to give detected helmets an ID and keep that ID across video frames.

The tracking result will be saved as a new video.

```python
# Run helmet detection and tracking on the video

results = model.track(
    source=video_path,
    tracker="bytetrack.yaml",
    imgsz=640,
    conf=0.25,
    persist=True,
    save=True
)
```

## Improving Tracking Stability

The first tracking test was able to assign IDs to detected helmets.

However, some IDs changed while the same rider was still visible. This happens when the helmet detection is temporarily lost between frames.

A slightly lower confidence threshold will be tested to give the tracker more detections and help it keep the same ID for longer.

```python
# Try tracking again with a lower confidence threshold

results = model.track(
    source=video_path,
    tracker="bytetrack.yaml",
    imgsz=640,
    conf=0.15,
    persist=True,
    save=True
)
```
